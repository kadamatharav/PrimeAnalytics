```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
# Load Dataset
title=pd.read_csv(r'/content/drive/MyDrive/Labmentix Projects/Amazon Prime EDA/titles.csv')
credits=pd.read_csv(r'/content/drive/MyDrive/Labmentix Projects/Amazon Prime EDA/credits.csv')
```

```python
# merging title and Credits dataset
df=pd.merge(title,credits,on='id')
```

```python
credits_grouped = credits.groupby("id").agg({
    "name": lambda x: ", ".join(x.astype(str)),
    "role": lambda x: ", ".join(x.astype(str))
}).reset_index()

df = pd.merge(
    title,
    credits_grouped,
    on="id",
    how="left"
)

print(df.shape)
df.head()
```

```python
df.to_csv("amazon_prime_merged.csv", index=False)

print("Merged dataset saved successfully!")
```

```python
# Convert genres into separate rows
genres = df["genres"].dropna().str.split(",").explode()

# Remove extra spaces
genres = genres.str.strip()

# Count genres
genre_counts = genres.value_counts()

genre_counts.head(10)
```

```python
plt.figure(figsize=(10,6))

genre_counts.head(10).plot(kind="bar")

plt.title("Top 10 Genres on Amazon Prime")
plt.xlabel("Genre")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)

plt.show()
```

####The platform offers content across many genres. Drama, Comedy, and International content are among the prominent categories, showing a diverse content library.
1. Why did you pick this chart?
A bar chart was selected because it makes it easy to compare the number of titles across different genres.

2. What insights were found?
Amazon Prime has a diverse content library covering multiple genres. Drama, Comedy, and other major categories represent a significant portion of the catalog.

3. Business impact?
Yes. Genre popularity can help Amazon Prime understand audience interests and plan future content acquisition and production.
