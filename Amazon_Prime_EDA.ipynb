### Import Libraries

```python
# Import Libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

### Dataset Loading

```python
# Load Dataset
title=pd.read_csv(r'/content/drive/MyDrive/Labmentix Projects/Amazon Prime EDA/titles.csv')
credits=pd.read_csv(r'/content/drive/MyDrive/Labmentix Projects/Amazon Prime EDA/credits.csv')
```

```python
from google.colab import drive
drive.mount('/content/drive')
```

```python
# merging title and Credits dataset
df=pd.merge(title,credits,on='id')
```

```python
credits_grouped = credits.groupby("id").agg({
    "name": lambda x: ", ".join(x.astype(str)),
    "role": lambda x: ", ".join(x.astype(str))
}).reset_index()

df = pd.merge(
    title,
    credits_grouped,
    on="id",
    how="left"
)

print(df.shape)
df.head()
```

```python
df.to_csv("amazon_prime_merged.csv", index=False)

print("Merged dataset saved successfully!")
```

```python
df = df.drop_duplicates()

print("Shape after removing duplicates:", df.shape)
```

```python
print("Type:")
print(df["type"].value_counts())

print("\nAge Certification:")
print(df["age_certification"].value_counts())

print("\nGenres:")
print(df["genres"].value_counts().head(10))
```

```python
df.describe()
```

```python
print("Minimum release year:", df["release_year"].min())
print("Maximum release year:", df["release_year"].max())
```

```python
plt.figure(figsize=(7,5))

sns.countplot(data=df, x="type")

plt.title("Movies vs TV Shows")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")

plt.show()
```

```python
# Convert genres into separate rows
genres = df["genres"].dropna().str.split(",").explode()

# Remove extra spaces
genres = genres.str.strip()

# Count genres
genre_counts = genres.value_counts()

genre_counts.head(10)
```

```python
plt.figure(figsize=(10,6))

genre_counts.head(10).plot(kind="bar")

plt.title("Top 10 Genres on Amazon Prime")
plt.xlabel("Genre")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)

plt.show()
```

```python
release_year = df["release_year"].value_counts().sort_index()

release_year.tail(20)
```

```python
plt.figure(figsize=(12,6))

release_year.plot(kind="line")

plt.title("Number of Titles by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.grid()
plt.show()
```

```python
most_titles_year = release_year.idxmax()
most_titles_count = release_year.max()

print("Year with most titles:", most_titles_year)
print("Number of titles:", most_titles_count)
```

```python
# Count the number of titles for each age certification/rating
rating_counts = df["age_certification"].value_counts()

# Display the ratings
rating_counts
```

```python
# Create a figure for the rating distribution
plt.figure(figsize=(10, 6))

# Create a bar chart of ratings
rating_counts.plot(kind="bar")

# Add title and axis labels
plt.title("Distribution of Content Ratings")
plt.xlabel("Rating")
plt.ylabel("Number of Titles")

# Rotate x-axis labels for better readability
plt.xticks(rotation=45)

# Display the graph
plt.show()
```

```python
# Find the rating with the highest number of titles
most_common_rating = rating_counts.idxmax()

# Find the number of titles having that rating
most_common_count = rating_counts.max()

# Display the result
print("Most common rating:", most_common_rating)
print("Number of titles:", most_common_count)
```

```python
# Check the first few runtime values
df["runtime"].head()
```

```python
# Calculate the average runtime of all titles
average_runtime = df["runtime"].mean()

# Display the average runtime
print("Average runtime:", round(average_runtime, 2), "minutes")
```

```python
# Find the longest runtime
longest_runtime = df["runtime"].max()

# Find the shortest runtime
shortest_runtime = df["runtime"].min()

# Display the results
print("Longest runtime:", longest_runtime, "minutes")
print("Shortest runtime:", shortest_runtime, "minutes")
```

```python
# Create a figure for the runtime distribution
plt.figure(figsize=(10, 6))

# Create a histogram of runtimes
plt.hist(df["runtime"].dropna(), bins=30)

# Add title and axis labels
plt.title("Distribution of Content Runtime")
plt.xlabel("Runtime (Minutes)")
plt.ylabel("Number of Titles")

# Display the graph
plt.show()
```

```python
# Check the first few IMDb scores
df["imdb_score"].head()
```

```python
# Calculate the average IMDb score
average_score = df["imdb_score"].mean()

# Display the average score
print("Average IMDb Score:", round(average_score, 2))
```

```python
# Find the highest IMDb score
highest_score = df["imdb_score"].max()

# Find the lowest IMDb score
lowest_score = df["imdb_score"].min()

# Display the results
print("Highest IMDb Score:", highest_score)
print("Lowest IMDb Score:", lowest_score)
```

```python
# Create a figure for the IMDb score distribution
plt.figure(figsize=(10, 6))

# Create a histogram of IMDb scores
plt.hist(df["imdb_score"].dropna(), bins=20)

# Add title and axis labels
plt.title("Distribution of IMDb Scores")
plt.xlabel("IMDb Score")
plt.ylabel("Number of Titles")

# Display the graph
plt.show()
```

```python
# Sort titles by IMDb score in descending order
top_10 = df.sort_values("imdb_score", ascending=False).head(10)

# Display title and IMDb score
top_10[["title", "imdb_score"]]
```

```python
# Check the first few IMDb vote values
df["imdb_votes"].head()
```

```python
# Calculate the correlation between IMDb score and IMDb votes
correlation = df["imdb_score"].corr(df["imdb_votes"])

# Display the correlation
print("Correlation between IMDb Score and IMDb Votes:", round(correlation, 2))
```

```python
# Create a figure for the scatter plot
plt.figure(figsize=(10, 6))

# Create a scatter plot
plt.scatter(df["imdb_votes"], df["imdb_score"], alpha=0.5)

# Add title and axis labels
plt.title("IMDb Score vs IMDb Votes")
plt.xlabel("IMDb Votes")
plt.ylabel("IMDb Score")

# Display the graph
plt.show()
```

```python
# Sort titles by IMDb votes in descending order
most_voted = df.sort_values("imdb_votes", ascending=False).head(10)

# Display the title, score, and votes
most_voted[["title", "imdb_score", "imdb_votes"]]
```

```python
# Select rows where the person is a director
directors = credits[credits["role"] == "DIRECTOR"]

# Count the number of titles directed by each person
top_directors = directors["name"].value_counts().head(10)

# Display the top 10 directors
top_directors
```

```python
# Create a figure for the top directors
plt.figure(figsize=(10, 6))

# Create a bar chart
top_directors.plot(kind="bar")

# Add title and axis labels
plt.title("Top 10 Directors by Number of Titles")
plt.xlabel("Director")
plt.ylabel("Number of Titles")

# Rotate names for better readability
plt.xticks(rotation=45)

# Display the graph
plt.show()
```

```python
# Select rows where the person is an actor
actors = credits[credits["role"] == "ACTOR"]

# Count the number of titles for each actor
top_actors = actors["name"].value_counts().head(10)

# Display the top 10 actors
top_actors
```

```python
# Create a figure for the top actors
plt.figure(figsize=(10, 6))

# Create a bar chart
top_actors.plot(kind="bar")

# Add title and axis labels
plt.title("Top 10 Actors by Number of Titles")
plt.xlabel("Actor")
plt.ylabel("Number of Titles")

# Rotate names for better readability
plt.xticks(rotation=45)

# Display the graph
plt.show()
```

```python
# Calculate the average IMDb score for Movies and TV Shows
average_score_type = df.groupby("type")["imdb_score"].mean()

# Display the average scores
average_score_type
```

```python
# Create a figure for the comparison
plt.figure(figsize=(8, 5))

# Create a bar chart
average_score_type.plot(kind="bar")

# Add title and axis labels
plt.title("Average IMDb Score: Movies vs TV Shows")
plt.xlabel("Content Type")
plt.ylabel("Average IMDb Score")

# Rotate x-axis labels
plt.xticks(rotation=0)

# Display the graph
plt.show()
```

```python
# Count the number of Movies and TV Shows
content_type = df["type"].value_counts()

# Display the counts
content_type
```

```python
# Select only Movies
movies = df[df["type"] == "MOVIE"]

# Sort Movies by IMDb score
top_movies = movies.sort_values("imdb_score", ascending=False).head(10)

# Display the top 10 Movies
top_movies[["title", "imdb_score"]]
```

```python
# Select only TV Shows
tv_shows = df[df["type"] == "SHOW"]

# Sort TV Shows by IMDb score
top_tv_shows = tv_shows.sort_values("imdb_score", ascending=False).head(10)

# Display the top 10 TV Shows
top_tv_shows[["title", "imdb_score"]]
```

```python
# Create a figure for the top Movies
plt.figure(figsize=(10, 6))

# Create a bar chart
plt.barh(top_movies["title"], top_movies["imdb_score"])

# Add title and axis labels
plt.title("Top 10 Highest-Rated Movies")
plt.xlabel("IMDb Score")
plt.ylabel("Movie")

# Display the graph
plt.show()
```

```python
# Create a figure for the top TV Shows
plt.figure(figsize=(10, 6))

# Create a bar chart
plt.barh(top_tv_shows["title"], top_tv_shows["imdb_score"])

# Add title and axis labels
plt.title("Top 10 Highest-Rated TV Shows")
plt.xlabel("IMDb Score")
plt.ylabel("TV Show")

# Display the graph
plt.show()
```

```python
# Split the production countries into separate values
countries = df["production_countries"].dropna().str.split(",").explode()

# Remove extra spaces
countries = countries.str.strip()

# Count the number of titles from each country
country_counts = countries.value_counts()

# Display the top 10 countries
country_counts.head(10)
```

```python
# Create a figure for the country analysis
plt.figure(figsize=(10, 6))

# Create a bar chart of the top 10 countries
country_counts.head(10).plot(kind="bar")

# Add title and axis labels
plt.title("Top 10 Countries by Number of Titles")
plt.xlabel("Country")
plt.ylabel("Number of Titles")

# Rotate country names for better readability
plt.xticks(rotation=45)

# Display the graph
plt.show()
```

```python
# Find the country with the highest number of titles
top_country = country_counts.idxmax()

# Find the number of titles from that country
top_country_count = country_counts.max()

# Display the result
print("Country with most titles:", top_country)
print("Number of titles:", top_country_count)
```
